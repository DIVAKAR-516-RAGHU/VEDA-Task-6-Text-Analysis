## Text Analysis

### Objective

Analyze a given text and calculate:

- Number of words
- Number of characters
- Number of vowels
- Repeated words and their frequency

### Tools Used

- Python
- Jupyter Notebook

```python
import re
from collections import Counter
```

## 1. Input Text

A sample paragraph is used for text analysis.

```python
text = """Artificial intelligence is transforming technology.
Artificial intelligence helps computers learn from data.
Python is widely used in artificial intelligence and data analysis.
Python makes data analysis simple and practical."""

print("Original Text:")
print(text)
```

## 2. Text Preprocessing

The text is converted to lowercase and individual words are extracted.

```python
clean_text = text.lower()

words = re.findall(r"\b[a-zA-Z]+\b", clean_text)

print("Words:")
print(words)
```

## 3. Word Count

The total number of words is calculated using `len()`.

```python
word_count = len(words)

print("Number of words:", word_count)
```

## 4. Character Count

The character count is calculated using Python's `len()` function.

```python
character_count = len(text)

character_count_without_spaces = len(
    re.sub(r"\s", "", text)
)

print("Number of characters:", character_count)
print("Characters without spaces:", character_count_without_spaces)
```

## 5. Vowel Count

The number of vowels in the text is calculated by checking
for the vowels a, e, i, o, and u.

```python
vowels = "aeiou"

vowel_count = sum(
    1 for char in clean_text
    if char in vowels
)

print("Number of vowels:", vowel_count)
```

## 6. Repeated Word Analysis

`Counter` is used to calculate the frequency of each word.

Words occurring more than once are identified as repeated words.

```python
word_frequency = Counter(words)

repeated_words = {
    word: count
    for word, count in word_frequency.items()
    if count > 1
}

print("Repeated Words:")

for word, count in sorted(repeated_words.items()):
    print(f"{word}: {count}")
```

## 7. Complete Text Analysis Summary

```python
print("===== TEXT ANALYSIS SUMMARY =====")
print("Word count:", word_count)
print("Character count:", character_count)
print("Characters without spaces:", character_count_without_spaces)
print("Vowel count:", vowel_count)

print("\nRepeated words:")

for word, count in sorted(repeated_words.items()):
    print(f"{word}: {count}")
```
